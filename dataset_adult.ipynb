```python
pd.set_option('display.max_rows', 250)

df2 = df[df['marital_status'] == 'Never-married']
df2['relationship'].value_counts()
# correspondance ok entre 'never married' et relationship

df3 = df[df['marital_status'] == 'Married-civ-spouse']
df3['relationship'].value_counts()
df3[(df3['relationship']=='Own-child') | (df3['relationship']=='Other-relative')].head(250)
# outliers entre married-civ-spoused et own-child(143)
df3 = df3[df3['relationship'] != 'Own-child']


df4 = df[df['marital_status'] == 'Married-spouse-absent']
df4['relationship'].value_counts()
# outliers eavec 'unmarried'(183) et 'ownchild'(61)
df4 = df4[~df4['relationship'].isin(['Own-child', 'Unmarried'])]


df5 = df[df['marital_status'] == 'Separated']
df5['relationship'].value_counts()
# outliers avec 'ownchild'(145)potentiellement à passer en other-relative
df5 = df5[df5['relationship'] != 'Own-child']


df6 = df[df['marital_status'] == 'Divorced']
df6['relationship'].value_counts()
# outliers 'own-child"(455)
#Si une personne a été mariée par le passé (Divorcée, Séparée, Veuve) 
# mais qu'elle est étiquetée Own-child, on la bascule en Other-relative
df6 = df6[df6['relationship'] != 'Own-child']


df7 = df[df['marital_status'] == 'Widowed']
df7['relationship'].value_counts()
# outliers 'own-child"(25)
df7 = df7[df7['relationship'] != 'Own-child']

df7
# environ mille lignes supprimées 
df_drop = pd.concat([df2,df3,df4,df5,df6,df7], ignore_index= True)
df_drop

```
